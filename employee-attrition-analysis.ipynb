```python
numeric_cols = df.select_dtypes(include="number").columns.tolist()

groupby_results = {}

for col in df.columns:
    # Skip high-cardinality columns such as employee IDs
    if df[col].nunique(dropna=False) <= 50:
        value_cols = [c for c in numeric_cols if c != col]

        grouped = df.groupby(col, dropna=False)

        summary = grouped.size().to_frame("row_count")

        if value_cols:
            aggregated = grouped[value_cols].agg(["mean", "median", "min", "max"])
            aggregated.columns = [
                f"{column}_{stat}" for column, stat in aggregated.columns
            ]
            summary = summary.join(aggregated)

        groupby_results[col] = summary

# Example: view the summary grouped by Attrition
display(groupby_results["Attrition"])
```

```python
plt.figure(figsize=(12, 6))
palette=['blue', 'orange', 'green', 'red', 'purple', 'brown', 'pink', 'gray', 'olive', 'cyan']
df.groupby('Age').size().plot(kind='bar', figsize=(12, 6), color=palette)
plt.title("Distribution of Employees by Age")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.savefig(VISUALIZATION_DIR / "age-distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Average monthly income by job role, sorted from highest to lowest
avg_salary_by_role = (
    df.groupby("JobRole")["MonthlyIncome"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))
avg_salary_by_role.plot(kind="bar", color="steelblue")
plt.title("Average Monthly Salary by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Average Monthly Salary")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
save_visualization("average-monthly-income-by-job-role")
plt.show()
```

```python
import seaborn as sns

pivot = df.pivot_table(
    index='Department',
    columns='JobSatisfaction',
    values='MonthlyIncome',
    aggfunc='mean'
)

sns.heatmap(pivot, annot=True, fmt='.0f', cmap='Blues')

plt.title('Average Salary by Department and Job Satisfaction')
save_visualization("average-salary-by-department-and-job-satisfaction")
plt.show()
```
